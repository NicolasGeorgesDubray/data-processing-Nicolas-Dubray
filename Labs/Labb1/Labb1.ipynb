##### Relationships between teaching resources and graduation outcomes

```python
# (h)
cols_sfratio_terminal_expend_books_gradrate = ["S.F.Ratio", "Terminal", "Expend_Log1pScaled", "Books_Log1pScaled", "Grad.Rate"]

# Pearson correlation coefficients.
corr_sfratio_terminal_expend_books_to_gradrate = college[cols_sfratio_terminal_expend_books_gradrate].corr()['Grad.Rate']
print(corr_sfratio_terminal_expend_books_to_gradrate)
```

```python
# Scatter matrix
_ = scatter_matrix(college[cols_sfratio_terminal_expend_books_gradrate], figsize=(10,10))
```

I wanted to look at relationships between quantitative teaching resource variables and graduation rates, going by the following columns:

- Expend : Instructional expenditure per student
- Terminal : Percent of faculty with terminal degree
- S.F.Ratio : Student/faculty ratio
- Books : Estimated book costs

Assumptions:
- "Books" represents estimated student expenditure on college course books
- "Terminal" accounts for the Ph.D. degrees accounted for in "Ph.D.".

Going by the Pearson correlation coefficients, there's
- a weak negative correlation (-0.306710) between S.F.Ratio and Grad.Rate.
- a weak positive correlation (0.289527) between Terminal and Grad.Rate.
- a moderate positive correlation (0.424733) between Expend and Grad.Rate.
- a negligible correlation (-0.005642) between Books and Grad.Rate.

The scatter matrix shows
- S.F.Ratio-to-Grad.Rate having a downwards slope (a negative trend)
- Terminal-to-Grad.Rate having a quite diffused scatter
- Expend-to-Grad.Rate having an upwards slope (a positive trend)
- Books-to-Grad.Rate having a vertical, streaked oval scatter
